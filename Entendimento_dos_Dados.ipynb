# Análise Estatísticas 

###### Identificar padrões de comportamento dos usuários, compreender a distribuição dos dados, comparar grupos e investigar possíveis relações entre variáveis.

```python
print(f"Média de idade: {df_clean["Age"].mean().round()}")
```

```python
print(f"Mediana da idade: {df_clean["Age"].median().round()}")
```

```python
print(f"Moda da idade: {df_clean["Age"].mode().round()}")
```

###### estatísticas Para horas assistidas 

```python
print(f"Média das Horas assistidas: {df_clean["Watch_Time_Hours"].mean().round()}")
```

```python
print(f"Mediana das horas assistidas: {df_clean["Watch_Time_Hours"].median().round()}")
```

```python
print(f"Moda das horas assistidas: {df_clean["Watch_Time_Hours"].mode().round()}")
```
